```python
#Initial String

s = "Skill Morph"
print (s.upper())
print (s.lower())
print (s.partition(' '))
print (s.replace('Morph', 'Master'))
print (s.find('Morph'))
print (s.rstrip())
print (s.split())
print (s.startswith('Skill'))
print (s.isnumeric())
print (s.index('Morph'))
```

```python
#Create a Set

student_id = {112, 114, 116, 118, 115}
print('Student_ID:', student_id)

Vowel_letters = {'a', 'e', 'i', 'o', 'u'}
print('Vowel_letters:', Vowel_letters)

mixed_set = {'Hello', 101, -2, 'Bye'}
print('mixed_set:', mixed_set)

```

```python
#create an empty set
empty_set = set()
empty_dictionary = {}

print ('Data type of empty_set:', type(empty_set))
print ('Data type of empty_dictionary:', type(empty_dictionary))

```

```python
#duplicate item in set

numbers = {1, 2, 3, 4, 5, 2, 3}
print(numbers)
```

```python
#add item and remote set item
numbers = {21, 34,54, 12}
print ('Initial set:', numbers)
numbers.add(32)
print ('Updated set:', numbers)

companies = {'apple', 'microsoft', 'google'}
print ('Initial set:', companies)
companies.remove('apple')
print ('Updated set:', companies)
```
