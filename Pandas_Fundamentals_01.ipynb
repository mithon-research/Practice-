```python
#grouping
sales_data = {
    'Region': ['North', 'South', 'East', 'North', 'South', 'East', 'West', 'West'],
    'Product': ['A', 'A', 'B', 'B', 'A', 'B', 'A', 'B'],
    'Sales': [100, 150, 200, 220, 250, 229, 409, 260],
    'Quantity':[10, 12, 39, 28, 49, 25, 20, 54]
}
sales_df = pd.DataFrame(sales_data)
print("Sales DataFrame:")
print(sales_df)

#Group by operation
region_sales = sales_df.groupby('Region')['Sales'].sum()
print("\nRegion-wise Sales:")
print(region_sales)
```
