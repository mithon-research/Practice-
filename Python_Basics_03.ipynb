```python
#short a list

my_list = input(" Enter element saparated by space: "). split()
my_list = [int(x) for x in my_list]

my_list.sort()
print("sorted list in ascending order:", my_list)

my_list.sort (reverse = True)
print("sorted list in descending order:", my_list)

```

```python
#Merge two lists


list1 = input("enter elements of the first list separated by space:").split()
list1 = [int(x) for x in list1]

list2 = input("enter elements of the second list separated by space:").split()
list2 = [int(x) for x in list2]

merged_list = list1 + list2

merged_list = list(set(merged_list))

print("merged list without duplicates:", merged_list)
```

```python
#Find the intersection

list1 = input("enter elements of the first list separated by space:").split()
list1 = [int(x) for x in list1]

list2 = input("enter elements of the second list separated by space:").split()
list2 = [int(x) for x in list2]

intersection = list(set(list1) & set(list2))

print("intersection of the two lists:", intersection)
```
